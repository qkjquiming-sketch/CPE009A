1. Distance Formula :
Make a program that would calculate the distance between two points given a list of coordinates. Use the distance formula.
coorindates_list = [(1,1), (2,3)]

````python
import math

def calculate_distance(point1, point2):
    x1, y1 = point1
    x2, y2 = point2
    distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    return distance

coordinates_list = [(1, 1), (2, 3)]

point_a = coordinates_list[0]
point_b = coordinates_list[1]
distance = calculate_distance(point_a, point_b)
print("The distance between ", point_a, " and ", point_b, " is: ", round(distance, 2))
````

2. Simple Word Filter :
For a given string input, replace all the words “stupid” with an asterisk * equal to the length of the string. The new
string value should be displayed with the asterisks.


````python
def simple_word_filter(input_string, words_to_filter_list):
    filtered_string = input_string
    for word_to_filter in words_to_filter_list:
        replacement = "*" * len(word_to_filter)
        filtered_string = filtered_string.replace(word_to_filter, replacement)
    return filtered_string

words_to_filter = [
    "stupid",
    "dumb",
    "silly",
    ]

input_str = input("Enter your sentence/s: ")
filtered_str = simple_word_filter(input_str, words_to_filter)
print("\n")
print(filtered_str)
````

3. Phonebook : Create a simple phonebook program that can read from a list of dictionaries. The program should be able to display a
person’s name, contact, and address based from a user input which is the id of the record.

````python
phonebook_records = [
    {
        'id': 1,
        'name': 'Eli',
        'contact': 'Number ni Eli',
        'address': 'Bahay ni Eli'
    },
    {
        'id': 2,
        'name': 'Wayne',
        'contact': 'Number ni Wayne',
        'address': 'Bahay ni Wayne'
    },
    {
        'id': 3,
        'name': 'Andie',
        'contact': 'Andie 2 Na si Andie',
        'address': 'Somewhere over the rainbow'
    }
]

print("Phonebook records created successfully.")
````

````python
search_id = int(input("Enter the ID to search: "))
print(f"Searching for ID: {search_id}")
````

````python
found = False
for record in phonebook_records:
    if record['id'] == search_id:
        print(f"\nRecord found for ID {search_id}:")
        print(f"Name: {record['name']}")
        print(f"Contact: {record['contact']}")
        print(f"Address: {record['address']}")
        found = True
        break

if not found:
    print(f"\nNo record found for ID {search_id}.")
````
