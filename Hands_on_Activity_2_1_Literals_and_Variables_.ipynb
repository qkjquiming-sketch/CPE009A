````python
PHOA = float(input("Enter Hands-on Activity grade: "))
PQUIZ = float(input("Enter Prelim Quiz grade: "))
PAS = float(input("Enter Prelim Assignment grade: "))
PCS = PHOA*.5 + PQUIZ*.3 + PAS*.2
print("Prelim CS is: ", PCS)
PEX = float(input("Enter Prelim Exam grade: "))
PG = PEX*.5 + PCS*.5
print("Prelim Grade is: ", "%.2f" % PG)
````

**Supplementary Activity**:

1. Test 3 students from the program you created.
2. The program should show the name of the student, the PRELIM, MIDTERM and FINAL grades.
3. Convert the final grade into the T.I.P.s numerical grade. Please refer to the grading system.


````python
for i in range(3):
    print(f"\n----- Student {i + 1} -----")
    student_name = input("Enter student's name: ")

    # Prelim Grades
    print("\n--- Prelim Period ---")
    PHOA = float(input("Enter Prelim Hands-on Activity grade: "))
    PQUIZ = float(input("Enter Prelim Quiz grade: "))
    PAS = float(input("Enter Prelim Assignment grade: "))
    PEX = float(input("Enter Prelim Exam grade: "))

    PCS = PHOA * 0.5 + PQUIZ * 0.3 + PAS * 0.2
    PG = PEX * 0.5 + PCS * 0.5

    # Midterm Grades
    print("\n--- Midterm Period ---")
    MHOA = float(input("Enter Midterm Hands-on Activity grade: "))
    MQUIZ = float(input("Enter Midterm Quiz grade: "))
    MAS = float(input("Enter Midterm Assignment grade: "))
    MEX = float(input("Enter Midterm Exam grade: "))

    MCS = MHOA * 0.5 + MQUIZ * 0.3 + MAS * 0.2
    # Midterm Grade (MG) calculation
    MG = (1/3) * PG + (2/3) * (MEX * 0.5 + MCS * 0.5)

    # Final Grades
    print("\n--- Final Period ---")
    FHOA = float(input("Enter Final Hands-on Activity grade: "))
    FQUIZ = float(input("Enter Final Quiz grade: "))
    FAS = float(input("Enter Final Assignment grade: "))
    FEX = float(input("Enter Final Exam grade: "))

    FCS = FHOA * 0.5 + FQUIZ * 0.3 + FAS * 0.2
    # Final Grade (FG) calculation
    FG = (1/3) * MG + (2/3) * (FEX * 0.5 + FCS * 0.5)

    if 99 <= FG <= 100:
        tip_grade = 1.00
    elif 96 <= FG <= 98:
        tip_grade = 1.25
    elif 93 <= FG <= 95:
        tip_grade = 1.50
    elif 90 <= FG <= 92:
        tip_grade = 1.75
    elif 87 <= FG <= 89:
        tip_grade = 2.00
    elif 84 <= FG <= 86:
        tip_grade = 2.25
    elif 81 <= FG <= 83:
        tip_grade = 2.50
    elif 78 <= FG <= 80:
        tip_grade = 2.75
    elif 75 <= FG <= 77:
        tip_grade = 3.00
    elif FG < 75:
        tip_grade = 5.00
    else:
        tip_grade = 5.00

    print(f"\n----- Grades for {student_name} -----")
    print(f"Prelim Grade: {PG:.2f}")
    print(f"Midterm Grade: {MG:.2f}")
    print(f"Final Grade: {FG:.2f}")
    print(f"9Numerical Grade: {tip_grade:.2f}")
````
