For Loop

```python
week = ["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
for x in week:
	print(x)
```

Break Statement

```python
#Example 1
week = ["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
for x in week:
    print(x)
    if x == "Thursday":
        break
```

```python
#Example 2
week = ["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
for x in week:
    if x == "Thursday":
        break
    print(x)
```

```python
#Example 3
week = ["Sunday","Monday","Tuesday","Wednesday","Thursday","Friday","Saturday"]
for x in week:
    if x == "Thursday":
        break
print(x)
```

Looping thorugh a String

```python
for x in "week":
    print(x)
```

The range() Function

```python
#Example 1
print('range(6)')
for x in range(6):
    print(x)
print('range(2,6)')
for x in range(2, 6):
    print(x)
```

```python
#Example 2
# for x in range(2 : Start , 6 : End, 2 : Increment)
for x in range(2, 6, 2):
    print(x)
```

Nested Loops

```python
#Example 1
adjective = ['red', 'big', 'tasty']
fruits = ['apple', 'banana', 'cherry']

for x in adjective:
    for y in fruits:
        print(x,y)
```

While Loop

```python
# Example 1
i = 1
while i < 6:
    print(i)
    i += 1  #i = i+1 : Assignment Operator
```

```python
# Break + While Loop
i = 1
while i < 6:
    print(i)
    if i == 3:
        break
    i += 1  #i = i+1 : Assignment Operator
```

Continue Statement

```python
i = 0
while i < 6:
    i += 1  #i = i+1 : Assignment Operator
    if i == 3:
        continue
    print(i)
```

```python
i = 0
while i < 6:
    i += 1
    if i != 3:
        continue
    print(i)
```

Else Statement

```python
i = 1
while i < 6:
    print(i)
    i += 1
else:
    print('i is no longer lass than 6')
```
