Question #23
```
Write a program that uses nested loops to print this pattern:

*
**
***
****
*****
****
***
**
*
```


````python
i = 1
ast = '*'

while i != 0:
    if i <= 6:
        for i in range(1, 6, 1):
            print(ast*i)
            i = i + 1
    if i >= 1:
        for i in range(4, 0, -1):
            print(ast*i)
            i = i - 1


````
