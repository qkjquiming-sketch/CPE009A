```python
def gradeMarks(grade):
    if grade >= 90:
        return "A+"
    elif grade >= 80:
        return "A"
    elif grade >= 70:
        return "B+"
    elif grade >= 60:
        return "B"
    elif grade >= 50:
        return "C"
    elif grade >= 40:
        return "D"
    else:
        return "F"

def calculateResult(percentage):
    if percentage >= 90:
        return "Outstanding"
    elif percentage >= 80:
        return "Excellent"
    elif percentage >= 70:
        return "Very Good"
    elif percentage >= 60:
        return "Good"
    elif percentage >= 40:
        return "Satisfactory"
    else:
        return "Needs Improvement"

# User Input
subCount = int(input("Enter the number of subjects: "))
if subCount <= 0:
    print("Invalid input. Please enter a valid number of subjects.")
else:
    for i in range(subCount):
        print(f"Subject {i+1}")
        grade = float(input("\tEnter the grade: "))
        percentage = grade # Simplified from (grade / 100) * 100
        print(f"\tPercentage: {percentage}%")
        print(f"\tResult: {calculateResult(percentage)}")
        print(f"\tGrade Mark: {gradeMarks(grade)}")
```
