```python
def termCrit(actNumber):
    rawScore = []
    totalScore = []

    for i in range(0, actNumber):
        while True:
            entry = input(f"  {i+1} │ (Score/Total): ")
            try:
                inputScore, inputTotal = map(float, entry.split('/'))
                if inputScore > inputTotal:
                    print("\n--- Invalid Scores: Score cannot be greater than Total. Please try again.---\n")
                else:
                    rawScore.append(inputScore)
                    totalScore.append(inputTotal)
                    break
            except ValueError:
                print("\n--- Invalid Input: Please enter score in 'Score/Total' format (e.g., 25/30). Please try again.---\n")

    if sum(totalScore) == 0:
        return 0
    grade = sum(rawScore) / sum(totalScore)
    grade = grade * 100
    return grade
```

```python
def termCS():
    print("--- Class Standing Components ---")

    print("• Hands-on Activities: ")
    termHOA = termCrit(int(input("How many HOA? ")))

    print("• Quizzes:")
    termQuiz = termCrit(int(input("How many quizzes? ")))

    print("• Assignment: ")
    termAss = termCrit(int(input("How many assignments? ")))

    grade = (0.5 * termHOA) + (0.3 * termQuiz) + (0.2 * termAss)
    return grade
```

```python
def semesterGrade():
    print("──────── PRELIM PERIOD ────────")
    prelimGrade = termGrade()

    print("\n──────── MIDTERM PERIOD ────────")
    midtermGrade = termGrade()
    finalMG = (1/3)*prelimGrade + (2/3)*midtermGrade

    print("\n──────── FINAL PERIOD ────────")
    finalGrade = termGrade()
    finalFG = (1/3)*finalMG + (2/3)*finalGrade

    return finalFG
```

```python
def termGrade():
    CSgrade = termCS()

    # Input for Exam Score, using the same robust method as termCrit
    while True:
        entry = input("• Exam (Score/Total): ")
        try:
            examRaw, examTotal = map(float, entry.split('/'))
            if examRaw > examTotal:
                print("\n--- Invalid Scores: Score cannot be greater than Total. Please try again.---\n")
            else:
                break
        except ValueError:
            print("\n--- Invalid Input: Please enter score in 'Score/Total' format (e.g., 25/30). Please try again.---\n")

    examGrade = (examRaw/examTotal) * 100

    grade = (0.5 * examGrade) + (0.5 * CSgrade)
    return grade
```

```python
# --- Driver Code / Testing Trial Cases ---
# TIP Grading Scheme (2025 - 2026)
def convertGrade(finalGrade):
    if finalGrade >= 99: return 1.00, "Excellent"
    elif finalGrade >= 96: return 1.25, "Very Good"
    elif finalGrade >= 93: return 1.50, "Good"
    elif finalGrade >= 90: return 1.75, "Satisfactory"
    elif finalGrade >= 87: return 2.00, "Fairly Satisfactory"
    elif finalGrade >= 84: return 2.25, "Approaching Satisfactory"
    elif finalGrade >= 81: return 2.50, "Moderately Satisfactory"
    elif finalGrade >= 78: return 2.75, "Pass"
    elif finalGrade >= 75: return 3.00, "Conditional Pass"
    else: return 5.00, "Fail"
```

```python
# Driver Code
# Trial 1:
print("\n--- Full Semester Grade Calculation ---\n")
final_percentage = semesterGrade()
grade, rating = convertGrade(final_percentage)

print(f"\n──────── FINAL GRADE CALCULATION ────────")
print(f"2Your Final Grades:")
print(f"Final Percentage: {final_percentage:.2f} %")
print(f"Equivalent Grade: {grade}")
print(f"Rating: {rating}")
```
