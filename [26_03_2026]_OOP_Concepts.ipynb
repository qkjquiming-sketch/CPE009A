```python
# Car's Class Name and Milaeage
class Car:
    def __init__(self, name, mileage):
        self.name = name
        self.mileage = mileage

# ==============================================================================

obj1 = Car("Toyota", 2202)
obj2 = Car("Subaru", 1200)

# ==============================================================================

print(obj1.name)
print(obj1.mileage, "km/l")

print(obj2.name)
print(obj2.mileage, "km/l")
```

```python
# Car's Class Name and Milage with Description
class Car:
    def __init__(self, name, milage):
        self.name = name
        self.mileage = milage

# ==============================================================================

obj1 = Car("Toyota", 2202)
obj2 = Car("Subaru", 1200)

# ==============================================================================

print(obj1.name, obj1.mileage)
print(obj2.name, obj2.mileage)

# ==============================================================================

def description(self):
    return f"The {self.name} car gives the mileage of {self.mileage} km/l."

# ==============================================================================

print(description(obj1))
print(description(obj2))
```

```python
#Animal's Features like name, habitat, and temperature
class Animals:
    def __init__(self, name, habitat, temp):
        self.name = name
        self.habitat = habitat
        self.temp = temp

# ==============================================================================

    def features(self):
        return f"The {self.name} is a {self.temp}-blooded animal that lives on {self.habitat}."

# ==============================================================================

Mammals = Animals("Dog", "Land", "Warm")
Birds = Animals("Sparrow", "Air", "Warm")
Reptiles = Animals("Snake", "Terrestrial", "Cold")

# ==============================================================================

# ===== List =====
print(Mammals.name, "->", Mammals.habitat)
print(Birds.name, "->", Birds.habitat)
print(Reptiles.name, "->", Reptiles.habitat)

print("\n")

# ===== Features =====
print(Mammals.features())
print(Birds.features())
print(Reptiles.features())
```

```python
import math

# ==============================================================================

class Shapes:
    def __init__(self, name, sides):
       self.name = name
       self.sides = sides

    def description(self):
        return f"The {self.name} has {self.sides} sides."

    def area(self, **dimensions):
        name = self.name.lower()
        # ===== Triangle =====
        if name == "triangle":
            return 0.5 * dimensions.get('base', 0) * dimensions.get('height', 0)
        # ===== Circle =====
        elif name == "circle":
            return math.pi * (dimensions.get('radius', 0) ** 2)
        # ===== Square =====
        elif name == "square":
            return dimensions.get('side', 0) ** 2
        # ===== Rectangle =====
        elif name == "rectangle":
            return dimensions.get('length', 0) * dimensions.get('width', 0)
        else:
            return 0

# ==============================================================================

obj1 = Shapes("Triangle", 3)
obj2 = Shapes("Circle", 1)
obj3 = Shapes("Square", 4)
obj4 = Shapes("Rectangle", 4)

# ==============================================================================

# ===== Triangle =====
print(obj1.description())
print(f"Area of {obj1.name}: {obj1.area(base=10, height=8)}.")

# ===== Circle =====
print(obj2.description())
area_c = obj2.area(radius=5)
print(f"Area of {obj2.name}: {area_c:.2f}")

# ===== Square =====
print(obj3.description())
print(f"Area of {obj3.name}: {obj3.area(side=10)}")

# ===== Rectangle =====
print(obj4.description())
print(f"Area of {obj4.name}: {obj4.area(length=5, width=10)}")
```
